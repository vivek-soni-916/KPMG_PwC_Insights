```python
# Helper function to detect and parse format
def parse_date(val):
    if re.match(r"\d{2}/\d{2}/\d{4}", val):  # dd/mm/yyyy
        return pd.to_datetime(val, dayfirst=True, errors='coerce')
    else:  # assume yyyy-mm-dd or yyyy-mm-dd hh:mm:ss
        return pd.to_datetime(val, dayfirst=True, errors='coerce')
```

```python
# Applying parser
df['date'] = df['ng-binding'].apply(parse_date)
```

```python

def extract_pdf_text(url):
    try:
        response = requests.get(url, timeout=10)
        if response.status_code == 200:
            pdf_file = BytesIO(response.content)
            doc = fitz.open(stream=pdf_file, filetype="pdf")
            text = ""
            for page in doc:
                text += page.get_text()
            return text.strip()
        else:
            return f"Failed: HTTP {response.status_code}"
    except Exception as e:
        return f"Error: {str(e)}"
```

```python
mask = march_df["pdf_content"].isna() | (march_df["pdf_content"].str.strip() == "")
```

```python
for idx in march_df[mask].index:
    link = march_df.at[idx, "collection__item-link href"]  # replace with your actual column name
    content = extract_pdf_text(link)
    march_df.at[idx, "pdf_content"] = content
```

## Pre-processing

##### Removing spaecs and punctuations

```python
def preprocess(text):
    text = re.sub(r'\s+', ' ', str(text))          # Remove extra spaces/newlines
    text = re.sub(r'[^\w\s]', '', text)            # Remove punctuation
    text = text.lower().strip()                    # Lowercase and strip
    return text
```

```python
combined_df['clean_content'] = combined_df['content'].apply(preprocess)
```

```python
combined_df
```

##### Adding content length

```python
combined_df['content_length'] = combined_df['clean_content'].apply(lambda x: len(str(x)))
```
